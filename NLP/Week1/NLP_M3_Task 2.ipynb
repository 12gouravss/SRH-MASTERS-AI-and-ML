```python
# TASK 2


import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download("stopwords")
nltk.download("wordnet")

movies = [
    {
        "title": "Galaxy Warriors",
        "description": "A space adventure with robots fighting aliens in distant galaxies"
    },
    {
        "title": "Healthy Kitchen",
        "description": "A guide to healthy food recipes and balanced diet meals"
    },
    {
        "title": "Robot Future",
        "description": "Advanced robots help humans explore Mars and outer space"
    },
    {
        "title": "Jungle Quest",
        "description": "Explorers travel through dangerous jungles searching for treasure"
    },
    {
        "title": "Food Paradise",
        "description": "Delicious healthy food and organic cooking ideas for families"
    },
    {
        "title": "Space Mission",
        "description": "Astronauts begin a risky mission across the universe with AI robots"
    },
    {
        "title": "Ocean Mystery",
        "description": "Scientists discover hidden creatures deep inside the ocean"
    },
    {
        "title": "Fitness Revolution",
        "description": "Workout plans and healthy lifestyle programs for beginners"
    },
    {
        "title": "AI Kingdom",
        "description": "Artificial intelligence controls a futuristic robotic city"
    },
    {
        "title": "Cooking Master",
        "description": "Professional chefs prepare tasty and healthy meals"
    }
]

# TEXT PREPROCESSING

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def preprocess(text):

    # Lowercase, Tokenization, Remove stopwords, Lemmatization
    text = text.lower()
    text= re.sub(r"[^\w\s]", "", text)
    words = text.split()
    words= [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)

processed_docs = []

for movie in movies:
    combined_text = movie["title"] + " " + movie["description"]
    processed_docs.append(preprocess(combined_text))

# TF-IDF VECTORIZATION

vectorizer = TfidfVectorizer()
tfidf_matrix= vectorizer.fit_transform(processed_docs)

# USER QUERY

query = "space adventure with robots"
# Preprocess query
processed_query = preprocess(query)

# Convert query into TF-IDF vector
query_vector = vectorizer.transform([processed_query])

# COSINE SIMILARITY

similarity_scores= cosine_similarity(query_vector, tfidf_matrix)

# GET TOP 3 RESULTS

top_indices = similarity_scores[0].argsort()[-3:][::-1]
print("\nTop 3 Recommended Movies:\n")
for index in top_indices:
    print("Title:", movies[index]["title"])
    print("Description:", movies[index]["description"])
    print("Similarity Score:", similarity_scores[0][index])
    print("-" * 50)
```
